```python


import pandas as pd

# Load the BLS Excel files
df_2023 = pd.read_excel("all_data_M_2023.xlsx")
df_2024 = pd.read_excel("all_data_M_2024.xlsx")

# Check that both files loaded correctly
print("2023 dataset shape:", df_2023.shape)
print("2024 dataset shape:", df_2024.shape)

# Check the column names
print(df_2023.columns.tolist())
print(df_2024.columns.tolist())
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# Select the four computing occupations
occupations = [
    "Software Developers",
    "Data Scientists",
    "Computer Programmers",
    "Computer Network Architects"
]

# Filter the 2023 and 2024 datasets for national data
data_2023 = df_2023[
    (df_2023["OCC_TITLE"].isin(occupations)) &
    (df_2023["AREA_TYPE"] == 1) &
    (df_2023["AREA_TITLE"] == "U.S.")
].copy()

data_2024 = df_2024[
    (df_2024["OCC_TITLE"].isin(occupations)) &
    (df_2024["AREA_TYPE"] == 1) &
    (df_2024["AREA_TITLE"] == "U.S.")
].copy()

# Remove duplicate occupation records, if any
data_2023 = data_2023.drop_duplicates(subset=["OCC_TITLE"])
data_2024 = data_2024.drop_duplicates(subset=["OCC_TITLE"])

# Display the results
print("2023 employment:")
display(data_2023[["OCC_TITLE", "TOT_EMP", "A_MEAN"]])

print("2024 employment:")
display(data_2024[["OCC_TITLE", "TOT_EMP", "A_MEAN"]])
```

```python

# Combine employment figures for both years
comparison = pd.merge(
    data_2023[["OCC_TITLE", "TOT_EMP"]],
    data_2024[["OCC_TITLE", "TOT_EMP"]],
    on="OCC_TITLE",
    suffixes=("_2023", "_2024")
)

# Convert employment values to numeric
comparison["TOT_EMP_2023"] = pd.to_numeric(
    comparison["TOT_EMP_2023"], errors="coerce"
)
comparison["TOT_EMP_2024"] = pd.to_numeric(
    comparison["TOT_EMP_2024"], errors="coerce"
)

# Create a grouped bar chart
comparison = comparison.dropna()

comparison.plot(
    x="OCC_TITLE",
    y=["TOT_EMP_2023", "TOT_EMP_2024"],
    kind="bar",
    figsize=(11, 6)
)

plt.title("Employment Comparison by Computing Occupation: 2023 vs 2024")
plt.xlabel("Computing Occupation")
plt.ylabel("Total Employment")
plt.xticks(rotation=25, ha="right")
plt.legend(["2023", "2024"])
plt.tight_layout()
plt.show()
```

This chart will compare the employment figures for each occupation between the two years.

```python

# Calculate the change in employment from 2023 to 2024
comparison["Change in Employment"] = (
    comparison["TOT_EMP_2024"] - comparison["TOT_EMP_2023"]
)

comparison["Percentage Change (%)"] = (
    comparison["Change in Employment"]
    / comparison["TOT_EMP_2023"] * 100
).round(2)

# Display the comparison table
display(comparison)

# Create a percentage change chart
plt.figure(figsize=(10, 6))

plt.bar(
    comparison["OCC_TITLE"],
    comparison["Percentage Change (%)"]
)

plt.title("Percentage Change in Employment by Occupation, 2023–2024")
plt.xlabel("Computing Occupation")
plt.ylabel("Employment Change (%)")
plt.xticks(rotation=25, ha="right")
plt.axhline(y=0, linewidth=1)
plt.tight_layout()
plt.show()
```

Interpretation

This visualisation examines the percentage change in employment across four computing occupations between May 2023 and May 2024. Positive values indicate an increase in employment, while negative values indicate a decrease. The results help identify which occupations experienced the greatest relative growth or decline during this period.

```python

# Prepare the 2023 employment data
pie_data = data_2023.copy()

pie_data["TOT_EMP"] = pd.to_numeric(
    pie_data["TOT_EMP"], errors="coerce"
)

pie_data = pie_data.dropna(subset=["TOT_EMP"])

# Combine any duplicate occupation records
pie_data = pie_data.groupby("OCC_TITLE", as_index=False)["TOT_EMP"].sum()

# Create the pie chart
plt.figure(figsize=(9, 7))

plt.pie(
    pie_data["TOT_EMP"],
    labels=pie_data["OCC_TITLE"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Share of Employment by Computing Occupation, USA, May 2023")
plt.axis("equal")
plt.show()
```

This pie chart illustrates the distribution of employment across four selected computing occupations in the United States in May 2023. Each slice represents an occupation's share of the combined employment total for these four occupations. The chart helps identify which occupation accounts for the largest and smallest shares of employment.
