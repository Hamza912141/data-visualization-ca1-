## Introduction

The Excel dataset contains US Occupational Employment and Wage Statistics (OEWS) for May 2023, published by the Bureau of Labor Statistics (BLS). It includes information on occupations, employment numbers, hourly and annual wages, and geographic areas such as states and metropolitan areas. In this assignment, I will use the dataset to create visualisations and analyse employment and wage patterns in AI, Data Science, and Software Development.

```python
import pandas as pd

file_path = "all_data_M_2023.xlsx"

df = pd.read_excel(file_path, sheet_name="All May 2023 data")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())

print("Column names:")
print(df.columns.tolist())
```

```python
print(df[["OCC_TITLE", "TOT_EMP", "A_MEAN", "A_MEDIAN"]].head(10))
```

```python
keywords = ["data scientist", "software developer", "computer", "artificial intelligence"]

mask = df["OCC_TITLE"].str.contains(
    "|".join(keywords), case=False, na=False
)

display(df.loc[mask, ["OCC_TITLE", "TOT_EMP", "A_MEAN", "A_MEDIAN"]].head(30))
```

```python
import matplotlib.pyplot as plt

occupations = [
    "Software Developers",
    "Data Scientists",
    "Computer Programmers",
    "Computer Network Architects"
]

employment = [1656880, 192710, 120370, 174100]

plt.figure(figsize=(10, 6))
plt.bar(occupations, employment)

plt.title("Employment by Computing Occupation, USA, May 2023")
plt.xlabel("Occupation")
plt.ylabel("Total Employment")
plt.xticks(rotation=30, ha="right")

plt.tight_layout()
plt.show()
```

Objective: Compare employment levels across four computing occupations.



Chart choice: A bar chart makes numerical comparisons between categories easy.
