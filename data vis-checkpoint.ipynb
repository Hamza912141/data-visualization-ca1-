```python
import matplotlib.pyplot as plt

occupations = [
    "Software Developers",
    "Data Scientists",
    "Computer Programmers",
    "Computer Network Architects"
]

employment = [1656880, 192710, 120370, 174100]
annual_wages = [138110, 119040, 107750, 133930]

plt.figure(figsize=(9, 6))

plt.scatter(employment, annual_wages)

for i, occupation in enumerate(occupations):
    plt.annotate(
        occupation,
        (employment[i], annual_wages[i]),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.title("Employment vs Annual Mean Wage, USA, May 2023")
plt.xlabel("Total Employment")
plt.ylabel("Annual Mean Wage ($)")
plt.tight_layout()
plt.show()
```

```python

# Select the four computing occupations
occupations = [
    "Software Developers",
    "Data Scientists",
    "Computer Programmers",
    "Computer Network Architects"
]

# Select only national-level records
comparison = df[
    (df["AREA"] == 99) &
    (df["OCC_TITLE"].isin(occupations))
].copy()

# Remove duplicate occupation records
comparison = comparison.drop_duplicates(
    subset=["OCC_TITLE"]
)

display(
    comparison[["OCC_TITLE", "TOT_EMP", "A_MEAN", "A_MEDIAN"]]
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.bar(
    comparison["OCC_TITLE"],
    comparison["A_MEAN"]
)

plt.title("Annual Mean Wages by Computing Occupation, USA, May 2023")
plt.xlabel("Computing Occupation")
plt.ylabel("Annual Mean Wage ($)")
plt.xticks(rotation=25, ha="right")

plt.tight_layout()
plt.show()
```

The chart compares annual mean wages across four computing occupations in the United States in May 2023. Software Developers have the highest mean wage among the four occupations, while Computer Programmers have the lowest.
