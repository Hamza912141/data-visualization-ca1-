```python
import matplotlib.pyplot as plt

occupations = [
    "Software Developers",
    "Data Scientists",
    "Computer Programmers",
    "Computer Network Architects"
]

employment = [1656880, 192710, 120370, 174100]
annual_wages = [138110, 119040, 107750, 133930]

plt.figure(figsize=(9, 6))

plt.scatter(employment, annual_wages)

for i, occupation in enumerate(occupations):
    plt.annotate(
        occupation,
        (employment[i], annual_wages[i]),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.title("Employment vs Annual Mean Wage, USA, May 2023")
plt.xlabel("Total Employment")
plt.ylabel("Annual Mean Wage ($)")
plt.tight_layout()
plt.show()
```

Objective: To examine the relationship between employment levels and annual mean wages across four computing occupations.

Why this chart? A scatter plot allows numerical relationships and patterns between two variables to be explored.

Finding: Look at the points and describe whether higher employment appears to correspond to higher or lower wages. Avoid claiming a general relationship from only four occupations.

Visualisation 4: Distribution of Annual Mean Wages

This visualisation examines how annual mean wages are distributed across selected computing-related occupations in the United States in May 2023.

```python
wages = pd.to_numeric(df["A_MEAN"], errors="coerce")
wages = wages.dropna()

print(wages.dtype)
print(wages.head())
```

```python
plt.figure(figsize=(9, 6))
plt.hist(wages.to_numpy(), bins=10, edgecolor="black")

plt.title("Distribution of Annual Mean Wages")
plt.xlabel("Annual Mean Wage ($)")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```
